```python
import pandas as pd
import re

df = pd.read_csv("data_cleaned.csv")

# hedges were sourced from a list from hyland's 1988 paper
hedges = [
    r"\babout\b", r"\balmost\b", r"\bapparent(ly)?\b", r"\bappear(s|ed|ing)?\b",
    r"\bassume(s|d|ing)?\b", r"\bbelieve(s|d|ing)?\b", r"\bcertain extent\b", r"\bcertain level\b",
    r"\bcertain amount\b", r"\bclaim(s|ed|ing)?\b", r"\bcommonly\b", r"\bconsider(s|ed|ing)?\b",
    r"\bcould\b", r"\bcouldn'?t\b", r"\bdepend(s|ed|ing)?\b", r"\bestimate(s|d|ing)?\b",
    r"\bfrequently\b", r"\bgenerally\b", r"\bguess(es|ed|ing)?\b", r"\bhypothesi[sz](e|es|ed|ing)?\b",
    r"\bimply(ing|ies|ied)?\b", r"\bindicate(s|d|ing)?\b", r"\binterpret(s|ed|ing)?\b",
    r"\blikely\b", r"\bmainly\b", r"\bmay\b", r"\bmight\b", r"\bmost\b", r"\bperhaps\b",
    r"\bplausible\b", r"\bpossible\b", r"\bpossibly\b", r"\bpostulate(s|d|ing)?\b",
    r"\bpresume(s|d|ing)?\b", r"\bprobabl(e|y)?\b", r"\bquite\b", r"\brather\b",
    r"\breported(ly)?\b", r"\broughly\b", r"\bsometimes\b", r"\bspeculate(s|d|ing)?\b",
    r"\bsuggest(s|ed|ing)?\b", r"\bsuppose(s|d|ing)?\b", r"\btend(s|ed|ing)?\b",
    r"\bthink(s|ing)?\b", r"\btypically\b", r"\buncertain(ly)?\b", r"\bunlikel(y)?\b",
    r"\busually\b", r"\bwould\b", r"\bseem(s|ed|ing)?\b", r"\bassumption(s)?\b", r"\bimagin(e|ed|ing)?\b"
]

# count hedges in each line
def count_hedges(text, patterns):
    if pd.isnull(text):
        return 0
    text = text.lower()
    return sum(len(re.findall(pattern, text)) for pattern in patterns)

df['hedge_count'] = df['text'].apply(lambda x: count_hedges(x, hedges))

# total number of words per line
df['word_count'] = df['text'].apply(lambda x: len(str(x).split()))

# associate with gender?
gender_group = df.groupby('gender').agg({
    'hedge_count': 'sum',
    'word_count': 'sum'
}).reset_index()

gender_group['hedges_per_1000_words'] = (
    gender_group['hedge_count'] / gender_group['word_count'] * 1000
)

print(gender_group[['gender', 'hedges_per_1000_words']])

gender_group.to_csv("hedging_by_gender.csv", index=False)
#^the above creates a new data set
```

```python
import pandas as pd
import re

df = pd.read_csv("data_cleaned.csv")

boosters = [
    r"\babsolutely\b", r"\bamazing\b", r"\balways\b", r"\bclearly\b", r"\bcompletely\b",
    r"\bdefinitely\b", r"\bdramatically\b", r"\bessential(ly)?\b", r"\bexceptionally\b",
    r"\bextremely\b", r"\bfundamental(ly)?\b", r"\bgreatly\b", r"\bhighly\b",
    r"\bhugely\b", r"\bimportant(ly)?\b", r"\bincredibly\b", r"\bindeed\b",
    r"\binvariably\b", r"\bis\b", r"\bmust\b", r"\bnever\b", r"\bnone\b", r"\bonly\b",
    r"\bprove(s|d)?\b", r"\bsignificant(ly)?\b", r"\bsurely\b", r"\btotally\b",
    r"\btruly\b", r"\bundeniable\b", r"\bundoubtedly\b", r"\bunquestionably\b",
    r"\bvery\b"
]

def count_boosters(text, patterns):
    if pd.isnull(text):
        return 0
    text = text.lower()
    return sum(len(re.findall(pattern, text)) for pattern in patterns)

df['booster_count'] = df['text'].apply(lambda x: count_boosters(x, boosters))

df['word_count'] = df['text'].apply(lambda x: len(str(x).split()))

# associate with gender?
gender_group = df.groupby('gender').agg({
    'booster_count': 'sum',
    'word_count': 'sum'
}).reset_index()

gender_group['boosters_per_1000_words'] = (
    gender_group['booster_count'] / gender_group['word_count'] * 1000
)

print(gender_group[['gender', 'boosters_per_1000_words']])

#gender_group.to_csv("boosters_by_gender.csv", index=False)
#^the above creates a new data set
```
